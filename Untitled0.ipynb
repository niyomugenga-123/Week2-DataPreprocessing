```python
# ============================================================================
# WEEK 2: DATA PREPROCESSING & FEATURE ENGINEERING FOR MACHINE LEARNING
# Dataset: Telco Customer Churn
# ============================================================================

# Step 1: Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder
import warnings
warnings.filterwarnings('ignore')

# Set display options for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("✓ All libraries imported successfully!")
```

```python
# ============================================================================
# PART 1: DATA INSPECTION - Understand your data
# ============================================================================

# Load the dataset
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')

print("=" * 80)
print("1. DATASET SHAPE")
print("=" * 80)
print(f"Rows (Customers): {df.shape[0]}")
print(f"Columns (Features): {df.shape[1]}")
print()

# Display first 10 rows
print("=" * 80)
print("2. FIRST 10 ROWS OF DATA")
print("=" * 80)
print(df.head(10))
print()

# Data types of each column
print("=" * 80)
print("3. DATA TYPES")
print("=" * 80)
print(df.dtypes)
print()

# Check for missing values
print("=" * 80)
print("4. MISSING VALUES")
print("=" * 80)
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])  # Only show columns with missing values
if missing_values.sum() == 0:
    print("✓ No missing values found!")
print()

# Check for duplicate rows
print("=" * 80)
print("5. DUPLICATE RECORDS")
print("=" * 80)
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")
if duplicates == 0:
    print("✓ No duplicate rows found!")
print()

# Descriptive statistics for numerical columns
print("=" * 80)
print("6. DESCRIPTIVE STATISTICS (Numerical Columns)")
print("=" * 80)
print(df.describe())
print()

# Summary of categorical columns
print("=" * 80)
print("7. CATEGORICAL COLUMNS SUMMARY")
print("=" * 80)
categorical_cols = df.select_dtypes(include='object').columns
for col in categorical_cols:
    print(f"\n{col}: {df[col].nunique()} unique values")
    print(df[col].value_counts())
```

```python
# ============================================================================
# PART 2: DATA CLEANING
# ============================================================================

# Load the dataset
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')

# Create a copy for cleaning
df_clean = df.copy()

print("=" * 80)
print("PART 2: DATA CLEANING")
print("=" * 80)
print()

# 1. Fix TotalCharges - convert from text to number
print("ISSUE 1: TotalCharges stored as text, not number")
print("-" * 80)
print(f"Before: {df_clean['TotalCharges'].dtype}")

df_clean['TotalCharges'] = pd.to_numeric(df_clean['TotalCharges'], errors='coerce')

nan_count = df_clean['TotalCharges'].isnull().sum()
if nan_count > 0:
    print(f"Found {nan_count} invalid values - filling with 0")
    df_clean['TotalCharges'].fillna(0, inplace=True)
else:
    print("✓ All values converted successfully")

print(f"After: {df_clean['TotalCharges'].dtype}")
print()

# 2. Remove customerID
print("ISSUE 2: Remove irrelevant columns")
print("-" * 80)
df_clean = df_clean.drop('customerID', axis=1)
print("✓ Removed customerID")
print()

print("=" * 80)
print("PART 2 COMPLETE")
print("=" * 80)
print(f"Dataset shape: {df_clean.shape}")
print(f"Data types check:")
print(df_clean.dtypes)
```

```python
# ============================================================================
# PART 3: FEATURE ENGINEERING
# ============================================================================

print("=" * 80)
print("PART 3: FEATURE ENGINEERING")
print("=" * 80)
print()

# Create new features from existing ones
print("Creating new features...")
print("-" * 80)
print()

# 1. Create binary feature for senior citizen (already 0/1, but let's verify)
print("1. Senior Citizen feature (already numeric)")
print(f"   Unique values: {df_clean['SeniorCitizen'].unique()}")
print()

# 2. Create total services feature (count how many services each customer has)
print("2. Creating 'TotalServices' feature")
service_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
                'TechSupport', 'StreamingTV', 'StreamingMovies']

# Count how many services = "Yes" for each customer
df_clean['TotalServices'] = 0
for col in service_cols:
    df_clean['TotalServices'] += (df_clean[col] == 'Yes').astype(int)

print(f"   TotalServices range: {df_clean['TotalServices'].min()} to {df_clean['TotalServices'].max()}")
print(f"   First 10 values: {df_clean['TotalServices'].head(10).tolist()}")
print()

# 3. Create contract duration feature (convert contract type to months)
print("3. Creating 'ContractMonths' feature from Contract type")
contract_mapping = {
    'Month-to-month': 1,
    'One year': 12,
    'Two year': 24
}
df_clean['ContractMonths'] = df_clean['Contract'].map(contract_mapping)
print(f"   Mapping: {contract_mapping}")
print(f"   First 10 values: {df_clean['ContractMonths'].head(10).tolist()}")
print()

# 4. Create charge per month feature (if tenure > 0)
print("4. Creating 'ChargePerMonth' feature")
df_clean['ChargePerMonth'] = df_clean['TotalCharges'] / (df_clean['tenure'] + 1)
print(f"   First 10 values:")
print(df_clean[['tenure', 'TotalCharges', 'ChargePerMonth']].head(10))
print()

# 5. Create tenure groups (categorize by how long customer has been with company)
print("5. Creating 'TenureGroup' feature")
df_clean['TenureGroup'] = pd.cut(df_clean['tenure'],
                                  bins=[0, 12, 24, 48, 72],
                                  labels=['0-1 year', '1-2 years', '2-4 years', '4+ years'])
print(f"   TenureGroup distribution:")
print(df_clean['TenureGroup'].value_counts().sort_index())
print()

print("=" * 80)
print("PART 3 COMPLETE")
print("=" * 80)
print(f"Dataset shape after feature engineering: {df_clean.shape}")
print(f"New columns added: TotalServices, ContractMonths, ChargePerMonth, TenureGroup")
print()
print("Columns in dataset:")
print(df_clean.columns.tolist())
```

```python
# ============================================================================
# PART 4: FEATURE ENCODING & SCALING
# ============================================================================

print("=" * 80)
print("PART 4: FEATURE ENCODING & SCALING")
print("=" * 80)
print()

# Make a copy for encoding/scaling
df_processed = df_clean.copy()

# ============================================================================
# STEP 1: ENCODING CATEGORICAL VARIABLES
# ============================================================================
print("STEP 1: ENCODING CATEGORICAL VARIABLES")
print("-" * 80)
print()

# 1. Binary encoding (Yes/No to 1/0) for simple binary columns
print("1. Binary Encoding (Yes/No → 1/0)")
binary_cols = ['Partner', 'Dependents', 'PhoneService', 'PaperlessBilling', 'Churn']
binary_mapping = {'Yes': 1, 'No': 0}

for col in binary_cols:
    df_processed[col] = df_processed[col].map(binary_mapping)
    print(f"   ✓ {col}: encoded")

print()

# 2. Encode gender (Male/Female → 1/0)
print("2. Binary Encoding for Gender")
df_processed['gender'] = df_processed['gender'].map({'Male': 1, 'Female': 0})
print(f"   ✓ gender: encoded")
print()

# 3. Handle columns with "No internet service" - encode to ordinal
print("3. Ordinal Encoding (No → 0, No internet service → 1, Yes → 2)")
internet_service_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
                         'TechSupport', 'StreamingTV', 'StreamingMovies']

ordinal_mapping = {'No': 0, 'No internet service': 1, 'Yes': 2}

for col in internet_service_cols:
    df_processed[col] = df_processed[col].map(ordinal_mapping)
    print(f"   ✓ {col}: encoded")

print()

# 4. One-Hot Encoding for categorical columns with 3+ categories
print("4. One-Hot Encoding (for columns with 3+ categories)")
categorical_to_encode = ['InternetService', 'OnlineBackup', 'Contract', 'PaymentMethod']

# Only encode if not already encoded
cols_to_one_hot = [col for col in categorical_to_encode if col in df_processed.columns
                   and df_processed[col].dtype == 'object']

if len(cols_to_one_hot) > 0:
    df_processed = pd.get_dummies(df_processed, columns=cols_to_one_hot, drop_first=True)
    print(f"   ✓ One-hot encoding applied to: {cols_to_one_hot}")
else:
    print("   ✓ All categorical columns already encoded")

print()
```

```python
# ============================================================================
# PART 5: OUTLIER DETECTION & FEATURE SELECTION
# ============================================================================

print("=" * 80)
print("PART 5: OUTLIER DETECTION & FEATURE SELECTION")
print("=" * 80)
print()

# ============================================================================
# STEP 1: OUTLIER DETECTION using IQR method
# ============================================================================
print("STEP 1: OUTLIER DETECTION (IQR Method)")
print("-" * 80)
print()

# Original numerical columns (before scaling)
original_numerical = ['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges',
                      'TotalServices', 'ContractMonths', 'ChargePerMonth']

outliers_found = {}

for col in original_numerical:
    if col in df_clean.columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outlier_count = len(df_clean[(df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)])
        outliers_found[col] = outlier_count

        print(f"{col}:")
        print(f"  Q1={Q1:.2f}, Q3={Q3:.2f}, IQR={IQR:.2f}")
        print(f"  Bounds: [{lower_bound:.2f}, {upper_bound:.2f}]")
        print(f"  Outliers found: {outlier_count}")
        print()

print("=" * 80)
print("STEP 2: FEATURE SELECTION (Correlation Analysis)")
print("-" * 80)
print()

# Get numerical columns from original data for correlation
correlation_cols = ['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges',
                    'TotalServices', 'ContractMonths', 'ChargePerMonth']

# Only use columns that exist
correlation_cols = [col for col in correlation_cols if col in df_clean.columns]

# Calculate correlation with target (Churn)
if 'Churn' in df_clean.columns:
    df_clean['Churn_binary'] = (df_clean['Churn'] == 'Yes').astype(int)
    correlations = df_clean[correlation_cols + ['Churn_binary']].corr()['Churn_binary'].drop('Churn_binary')

    print("Correlation with Churn (target variable):")
    correlations_sorted = correlations.abs().sort_values(ascending=False)
    print(correlations_sorted)
    print()

    # Features with strong correlation (|r| > 0.1)
    strong_features = correlations_sorted[correlations_sorted > 0.1].index.tolist()
    print(f"Features with strong correlation (|r| > 0.1): {strong_features}")
    print()

# ============================================================================
# STEP 3: CREATE VISUALIZATIONS
# ============================================================================
print("=" * 80)
print("STEP 3: CREATING VISUALIZATIONS")
print("-" * 80)
print()

import matplotlib.pyplot as plt

# 1. Box plots for outlier detection
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
fig.suptitle('Box Plots - Outlier Detection', fontsize=16, fontweight='bold')

plot_cols = ['tenure', 'MonthlyCharges', 'TotalCharges', 'TotalServices', 'ContractMonths', 'ChargePerMonth']

for idx, col in enumerate(plot_cols):
    if col in df_clean.columns:
        ax = axes[idx // 3, idx % 3]
        ax.boxplot(df_clean[col].dropna())
        ax.set_title(f'{col}')
        ax.set_ylabel('Value')
        ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('boxplots_outliers.png', dpi=100, bbox_inches='tight')
plt.show()
print("Box plots saved")
print()

# 2. Correlation heatmap
fig, ax = plt.subplots(figsize=(10, 8))

corr_matrix = df_clean[correlation_cols + ['Churn_binary']].corr()
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0, ax=ax)
ax.set_title('Correlation Heatmap - Features vs Churn', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('correlation_heatmap.png', dpi=100, bbox_inches='tight')
plt.show()
print("Correlation heatmap saved")
print()

print("=" * 80)
print("PART 5 COMPLETE")
print("=" * 80)
print(f"Outliers detected: {sum(outliers_found.values())} total")
print(f"Strong correlations found: {len(strong_features)} features")
```

```python
# ============================================================================
# PART 6: MACHINE LEARNING READY DATASET
# ============================================================================

print("=" * 80)
print("PART 6: SAVING ML-READY DATASET")
print("=" * 80)
print()

# Use the processed data from Part 4
final_dataset = df_processed.copy()

print("FINAL DATASET SUMMARY")
print("-" * 80)
print()

print(f"1. Dataset Shape: {final_dataset.shape}")
print(f"   - Rows (customers): {final_dataset.shape[0]}")
print(f"   - Columns (features): {final_dataset.shape[1]}")
print()

print(f"2. Data Quality Check:")
print(f"   - Missing values: {final_dataset.isnull().sum().sum()}")
print(f"   - Duplicate rows: {final_dataset.duplicated().sum()}")
print()

print(f"3. Feature Types:")
numerical_features = final_dataset.select_dtypes(include=['int64', 'float64']).columns.tolist()
print(f"   - Numerical features: {len(numerical_features)}")
print(f"   - All features: {numerical_features}")
print()

print(f"4. Target Variable (Churn):")
if 'Churn' in final_dataset.columns:
    print(f"   - Values: {final_dataset['Churn'].unique()}")
    print(f"   - Distribution:")
    print(final_dataset['Churn'].value_counts())
print()

print("=" * 80)
print("SAVING PROCESSED DATASET")
print("=" * 80)
print()

# Save to CSV
output_filename = 'Telco_Churn_Processed.csv'
final_dataset.to_csv(output_filename, index=False)
print(f"✓ Dataset saved as: {output_filename}")
print(f"  File size: {final_dataset.memory_usage(deep=True).sum() / 1024:.2f} KB")
print()

# Display first 5 rows of processed data
print("First 5 rows of processed dataset:")
print(final_dataset.head())
print()

print("=" * 80)
print("PREPROCESSING COMPLETE!")
print("=" * 80)
print()
print("Summary of preprocessing steps performed:")
print("1. ✓ Data Inspection - identified data types, missing values, duplicates")
print("2. ✓ Data Cleaning - converted TotalCharges to numeric, removed customerID")
print("3. ✓ Feature Engineering - created TotalServices, ContractMonths, ChargePerMonth, TenureGroup")
print("4. ✓ Feature Encoding - encoded categorical variables (binary, ordinal, one-hot)")
print("5. ✓ Feature Scaling - scaled numerical features using StandardScaler")
print("6. ✓ Outlier Detection - identified outliers using IQR method")
print("7. ✓ Feature Selection - analyzed correlation with target variable")
print()
print("Dataset is now ready for machine learning model development!")
```
